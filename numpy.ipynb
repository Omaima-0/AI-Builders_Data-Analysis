# Dot Product 
The main two rules for dot product to remember are:
1. The inner dimention must match:
* (3,2) @ (3,2) won't work
* (2,3) @ (2,3) will work 
* (3,2) @ (2,3) will work 

2.The resulting matrix has the shape of the outer dimetions:
* (2,3) @ (3,2) -> (2,2)
* (3,2) @ (2,3) -> (3,3)

NOTE: In NumPy, np.dot() and @ can be used to achive the same result for 1-2 dimention arrays. However, their behaviour begind to differ at arrays with 3 + dimenstions.

```python
np.random.seed(0)   # .seed(0) t avoid changing the result 
mat1 =np.random.randint(10, size = (3,3))
mat2 =np.random.randint(10, size = (3,2))

mat1.shape , mat2.shape
```

```python
mat1
```

```python
mat2
```

```python
np.dot(mat1,mat2)
```

```python
mat1 @ mat2      # same result
```
