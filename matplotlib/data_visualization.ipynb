```python
import pandas as pd 
```

```python
car_sales = pd.read_sv("../Pandas/car-sales.csv")
```

```python
# remove the $, comma , and dot 
car_sales["Price"] = car_sales["Price"].str.replace('[$,.]','', regex=True)
car_sales
```

```python
#remove last two zeros 
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
#adding a new coloumn
car_sales["Sales Date"] = pd.date_range("1/1/2026",periods =len(car_sales))
car_sales
```

```python
# create total sales based on the Price
car_sales["Total Sales"] = car_sales["Price"].astype(int).cumsum()
car_sales
```

```python
#create a plot for total sales and sales date 
car_sales.plot(x='Sales Date', y='Total Sales')
```

```python
#create a stater plot 
car_sales["Price"] = car_sales["Price"].astype(int)
car_sales.plot(x= "Odometer (KM)", y="Price", kind="scatter") # kind is the type of plot 
```

```python
#create bar plot 
car_sales.plot(x='Make', y ='Odometer (KM)',kind='bar')
```

```python
#first way to make histogram plot 
car_sales['Odometer (KM)'].plot.hist()
```

```python
#second way to make histogram plot 
car_sales['Odometer (KM)'].plot(kind='hist')
```

```python
# bins : every 10 counts as a group
car_sales['Odometer (KM)'].plot.hist(bins=30)
```

```python
car_sales["Price"].plot.hist(bins=20)
```
