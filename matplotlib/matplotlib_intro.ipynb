# Type of Plots using NumPy arrays

Most of figuring out what kind of plot to use is getting a feel for the data, then seeing what kind of plot suits best.


Matplotlib visualizations are built on NumPy arrays. 


Types of plots:
* line plot - ax.plot()
* Scatter plot - ax.scatter()
* Bar plot - ax.hist()
* Histogram plot - ax.hist()

We'll see how all these can be created as a method from matplotlob.pyplot.subplots()

```python
import numpy as np
```

```python
x = np.linspace(0,10,100)
x
```

```python
# to turn it as plot 
fig, ax = plt.subplots()
ax.plot(x, x**2)
```

```python
# to create scatter line    # we use it if I have different of individual points and they interact without connec
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

```python
# Bar plots: we use when we have different amount, catogaries, gender 
nm_of_sudents = {"uni students":10,
                 "High School Students": 20,
                 "Fresh Graduate": 15}
fig, ax = plt.subplots()
ax.bar(nm_of_sudents.keys(), nm_of_sudents.values())
```

```python
# Bar plots: we use when we have different amount, catogaries, gender 
nm_of_sudents = {"uni students":10,
                 "High School Students": 20,
                 "Fresh Graduate": 15}
fig, ax = plt.subplots()
ax.barh(nm_of_sudents.keys(), nm_of_sudents.values())
```

```python
#Histogram 
x = np.random.randn(1000)
fig,ax = plt.subplots()
ax.hist(x)
```

```python
#Histogram 
x = np.random.random(1000)
fig,ax = plt.subplots()
ax.hist(x)
```

```python
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,5))   #rows ,coloumns then figure size 
ax1.plot(x, x**2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nm_of_sudents.keys(), nm_of_sudents.values())
ax4.hist(np.random.randn(1000))
```
